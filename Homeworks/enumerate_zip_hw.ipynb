```python
# ============================================================
# Homework | Python -- enumerate() and zip()
# ============================================================
# Rules:
#   - Fill in every ... to make the script produce the
#     expected output shown in the comments below each block.
#   - Run the whole file at once -- if it runs clean with
#     correct output, you are done.
#   - Hints are at the very bottom. Do not scroll down until
#     you are genuinely stuck.
# 
# zip() returns an iterator object, and the elements yielded by that iterator are tuples. 
# ============================================================


# ------------------------------------------------------------
# Q1 [Simple] -- API response logger
# A list of API endpoint names just returned responses.
# Print each one with a 1-based sequence number so the log
# is human-readable.
# ------------------------------------------------------------

endpoints = ['/health', '/auth/login', '/portfolio', '/holdings', '/prices']

for idx, val in enumerate(endpoints, start=1):
    print(f'{idx}. {val}')

# expected:
# 1. /health
# 2. /auth/login
# 3. /portfolio
# 4. /holdings
# 5. /prices


# ------------------------------------------------------------
# Q2 [Simple] -- Stock ticker lookup table
# Build a dict mapping each stock symbol to its price using
# zip. One line only.
# ------------------------------------------------------------

symbols = ['INFY', 'TCS', 'HDFC', 'RELIANCE']
prices  = [1450.0, 3800.5, 1600.0, 2900.75]

ticker_map = {s:p for s, p in zip(symbols, prices)} 
#ticker_map = dict(zip(symbols, prices))

print(ticker_map)
# expected:
# {'INFY': 1450.0, 'TCS': 3800.5, 'HDFC': 1600.0, 'RELIANCE': 2900.75}


# ------------------------------------------------------------
# Q3 [Simple] -- Find the first config key with a None value
# Iterate through a config dict and use enumerate to find the
# index and key of the first entry whose value is None.
# ------------------------------------------------------------

config = {
    'host': 'localhost',
    'port': 5432,
    'db_name': None,
    'user': 'admin',
    'password': None,
}

first_missing_idx  = None
first_missing_key  = None

for idx, val in enumerate(config.items()):
    if val[1] == None:
        first_missing_idx = idx
        first_missing_key = val
        break

print(first_missing_idx, first_missing_key)
# expected:
# 2 db_name


# ------------------------------------------------------------
# Q4 [Simple] -- Dot product of two portfolio weight vectors
# Given two vectors, compute their dot product using zip.
# No numpy allowed.
# ------------------------------------------------------------

weights  = [0.4, 0.3, 0.2, 0.1]
returns  = [0.12, 0.08, 0.15, 0.05]

portfolio_return = sum([i[0]*i[1] for i in zip(weights, returns)])

print(round(portfolio_return, 4))
# expected:
# 0.107


# ------------------------------------------------------------
# Q5 [Simple] -- Replace failing health-check entries in place
# A list of service statuses is returned from a monitoring
# call. Use enumerate to replace every 'DOWN' entry with
# 'RESTARTING' directly in the list.
# ------------------------------------------------------------

statuses = ['UP', 'DOWN', 'UP', 'DOWN', 'UP', 'DOWN']

for idx, val in enumerate(statuses):
    if val == 'DOWN':
        statuses[idx] = 'RESTARTING'
    
print(statuses)
# expected:
# ['UP', 'RESTARTING', 'UP', 'RESTARTING', 'UP', 'RESTARTING']


# ------------------------------------------------------------
# Q6 [Medium] -- Ranked portfolio holdings by value
# Zip two lists together, sort by holding value descending,
# then use enumerate to print a numbered leaderboard.
# ------------------------------------------------------------

holdings = ['INFY', 'TCS', 'HDFC', 'RELIANCE', 'WIPRO']
values   = [145000, 380000, 160000, 290000, 95000]

ranked = sorted(list(zip(holdings, values)), reverse=True, key=lambda x:x[1])

for idx, val in enumerate(ranked, start=1):
    print(f'Rank {idx} | {val[0]}\t|{val[1]}')

# expected:
# Rank 1 | TCS       | 380000
# Rank 2 | RELIANCE  | 290000
# Rank 3 | HDFC      | 160000
# Rank 4 | INFY      | 145000
# Rank 5 | WIPRO     | 95000


# ------------------------------------------------------------
# Q7 [Medium] -- Detect price anomalies between two feeds
# Two data feeds return prices for the same symbols. Use zip
# and enumerate to find every index where the two feeds
# disagree by more than a 1.0 tolerance, and print the
# index, symbol, and both prices.
# ------------------------------------------------------------

symbols_feed = ['INFY', 'TCS', 'HDFC', 'RELIANCE', 'WIPRO']
feed_a       = [1450.0, 3800.5, 1600.0, 2900.75, 420.0]
feed_b       = [1451.0, 3810.0, 1600.0, 2898.50, 419.5]

for i, v in enumerate(zip(symbols_feed, feed_a, feed_b)):
    if abs(v[1] - v[2]) > 1:
        print(f'idx={i}\t| {v[0]}\t| feed_a={v[1]}\t| feed_b={v[2]}')

# expected:
# idx=1 | TCS    | feed_a=3800.5 | feed_b=3810.0
# idx=3 | RELIANCE | feed_a=2900.75 | feed_b=2898.5


# ------------------------------------------------------------
# Q8 [Medium] -- Unzip, apply a charge, re-zip
# A list of (transaction_id, amount) tuples comes from a
# payments API. Unzip them, apply a 2% processing fee to
# every amount, then re-zip and print the updated records.
# ------------------------------------------------------------

transactions = [
    ('TXN001', 10000),
    ('TXN002', 25000),
    ('TXN003', 5000),
    ('TXN004', 80000),
]

#important to remember - zip(*data) for unzipping. unzipping returns tuples, hence need to use map to make it list
tx_ids, amounts = map(list, zip(*transactions))

charged = [x*(1.02) for x in amounts]

for elem in zip(tx_ids, charged):
    print(f'{elem[0]}\t|{elem[1]}')

# expected:
# TXN001 | 10200.0
# TXN002 | 25500.0
# TXN003 | 5100.0
# TXN004 | 81600.0


# ------------------------------------------------------------
# Q9 [Complex] -- Build a field-diff report between two
# versions of a holdings record
# Two dicts represent old and new states of a holding.
# Use zip and enumerate on their items() to produce a
# structured diff -- only report fields that changed.
# When you pass two dictionaries directly into zip(), Python iterates over their keys.
# dict1 = {'a': 1, 'b': 2}
# dict2 = {'x': 10, 'y': 20}
# result = list(zip(dict1, dict2))
# Output: [('a', 'x'), ('b', 'y')]
# result = list(zip(dict1.values(), dict2.values()))
# Output: [(1, 10), (2, 20)]
# result = list(zip(dict1.items(), dict2.items()))
# Output: [(('a', 1), ('x', 10)), ((('b', 2), ('y', 20)))]
# ------------------------------------------------------------

old_holding = {
    'symbol': 'INFY',
    'qty': 100,
    'avg_price': 1400.0,
    'sector': 'IT',
    'active': True,
}

new_holding = {
    'symbol': 'INFY',
    'qty': 150,
    'avg_price': 1450.0,
    'sector': 'IT',
    'active': True,
}

diff = []

for idx, val in enumerate(zip(old_holding.items(), new_holding.items())):
    if val[0][1] != val[1][1]:
        diff.append({'field': val[0][0], 'old': val[0][1], 'new': val[1][1], 'idx':idx})

for entry in diff:
    print(entry)

# expected:
# {'field': 'qty',       'old': 100,    'new': 150,    'idx': 1}
# {'field': 'avg_price', 'old': 1400.0, 'new': 1450.0, 'idx': 2}


# ------------------------------------------------------------
# Q10 [Complex] -- Rolling 3-period return calculator
# Given a list of daily NAV (Net Asset Value) prices for a
# fund, use zip and enumerate to compute the rolling 3-day
# return at every valid window and collect any window where
# the return exceeds a 2% threshold into an alert list.
# Do not use any external library.


# # Create an iterator, then pass it 3 times to zip
# nums = [10, 20, 30, 40, 50] #60
# it = iter(nums)
# result = list(zip(it, it, it))
# Output: [(10, 20, 30), (40, 50, 60)]
# if 60 not there then [(10, 20, 30)]

# Zip the list with itself shifted by 1 and 2 positions
# result = list(zip(nums, nums[1:], nums[2:]))
# Output: [(10, 20, 30), (20, 30, 40), (30, 40, 50)]
# ------------------------------------------------------------

nav_prices = [100.0, 101.5, 103.2, 102.8, 106.0, 105.5, 109.3, 108.0]

threshold   = 0.02
alerts      = []

# hint for structure: zip three offset views of the list to get
# (day0, day1, day2) windows, then enumerate to get the window index



for i, (day0, day1, day2) in enumerate(zip(nav_prices, nav_prices[1:], nav_prices[2:])):
    rolling_return = (day2 - day0)/day0
    if rolling_return > threshold:
        alerts.append({'window': i, 'start_nav': day0, 'end_nav': day2, 'return': round(rolling_return, 4)})

for alert in alerts:
    print(alert)

# expected (returns rounded to 4dp):
# {'window': 0, 'start_nav': 100.0,  'end_nav': 103.2, 'return': 0.032}
# {'window': 2, 'start_nav': 103.2,  'end_nav': 106.0, 'return': 0.0271}
# {'window': 4, 'start_nav': 106.0,  'end_nav': 109.3, 'return': 0.0311}


# ============================================================
# HINTS -- only read if genuinely stuck
# ============================================================
#
# Q1: enumerate(endpoints, start=1) removes the need to add 1
#     manually. f"{i}. {ep}" formats the output.
#
# Q2: dict(zip(symbols, prices)) -- zip produces (key, value)
#     pairs, dict() accepts exactly that.
#
# Q3: enumerate(config.items()) gives (idx, (key, val)).
#     Unpack as: for i, (k, v) in enumerate(config.items())
#     Break as soon as you find the first None.
#
# Q4: sum(w * r for w, r in zip(weights, returns))
#     Each pair multiplied, all summed in one expression.
#
# Q5: for i, s in enumerate(statuses): if s == 'DOWN': statuses[i] = 'RESTARTING'
#
# Q6: sorted(zip(holdings, values), key=lambda x: x[1], reverse=True)
#     gives sorted (symbol, value) pairs. Then enumerate with start=1.
#     f"Rank {rank} | {sym:<10} | {val}" -- use :<10 to left-align the symbol.
#
# Q7: for i, (sym, a, b) in enumerate(zip(symbols_feed, feed_a, feed_b))
#     Check abs(a - b) > 1.0 as the anomaly condition.
#
# Q8: tx_ids, amounts = zip(*transactions) -- unpacks list of tuples.
#     charged = [a * 1.02 for a in amounts]
#     Then zip(tx_ids, charged) to re-pair them.
#
# Q9: zip(old_holding.items(), new_holding.items()) pairs up fields.
#     Unpack as: for i, ((k1, v1), (k2, v2)) in enumerate(zip(...))
#     If v1 != v2, append {'field': k1, 'old': v1, 'new': v2, 'idx': i}
#
# Q10: zip three sliced views -- zip(nav_prices, nav_prices[1:], nav_prices[2:])
#      gives (day0, day1, day2) per window.
#      rolling_return = round((d2 - d0) / d0, 4)
#      Append {'window': i, 'start_nav': d0, 'end_nav': d2, 'return': rolling_return}
#
# ============================================================
```
